```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from scipy import stats
matplotlib.style.use('ggplot')
%matplotlib inline
```

```python
town_1959_df.set_index(u'номер')
```

```python
plt.hist(np.log10(town_1959_df[u'население']))
```

```python
albuquerque_home_prices_df = pd.read_csv('Albuquerque Home Prices_data.txt', sep = '\t')
```

```python
albuquerque_home_prices_df = albuquerque_home_prices_df.replace(-9999, np.nan)
```

```python
albuquerque_home_prices_df.head()
```

```python
x = albuquerque_home_prices_df[albuquerque_home_prices_df['COR'] == 1]['PRICE']
y = albuquerque_home_prices_df[albuquerque_home_prices_df['COR'] == 0]['PRICE']
x.name, y.name = 'Corner', 'Not Corner'
```

```python
def two_histograms(x, y):
    x.hist(alpha = 0.5, weights = [1. / len(x)] * len(x))
    y.hist(alpha = 0.5, weights = [1. / len(y)] * len(y))
    plt.axvline(x.mean(), color = 'red', alpha = 0.8, linestyle = 'dashed')
    plt.axvline(y.mean(), color = 'blue', alpha = 0.8, linestyle = 'dashed')
    plt.legend([x.name, y.name])
```

```python
two_histograms(x, y)
```

```python
res = stats.mannwhitneyu(x, y)
print('p-value (by the Mann-Whitney-Wilcoxon test):', res[1])
```

```python
x = albuquerque_home_prices_df[albuquerque_home_prices_df['NE'] == 1]['PRICE']
y = albuquerque_home_prices_df[albuquerque_home_prices_df['NE'] == 0]['PRICE']
x.name, y.name = 'NE', 'Not NE'
```

```python
two_histograms(x, y)
```

```python
res = stats.median_test(x, y)
print('p-value (median test):', res[1])
```

```python
age_european_aristocrats_die_df = pd.read_csv('agedeath.dat.txt', sep = '\s+', header = None, names = ['group', 'age', 'index'])
```

```python
age_european_aristocrats_die_df.head()
```

```python
x = age_european_aristocrats_die_df[age_european_aristocrats_die_df['group'] == 'sovr']['age']
y = age_european_aristocrats_die_df[age_european_aristocrats_die_df['group'] == 'aris']['age']
x.name, y.name = 'sovr', 'aris'
```

```python
two_histograms(x, y)
```

```python
res = stats.fligner(x, y)
print('p-value (by the Fligner-Killin test):', res[1])
```

```python
res = stats.ttest_ind(x, y, equal_var = False)
print('p-value (by the Pearson`s  test):', res[1])
```

```python
interference_df = pd.read_csv('interference.csv')
```

```python
interference_df.head()
```

```python
x = interference_df['DiffCol']	
y = interference_df['Black']
x.name, y.name = 'DiffCol', 'Black'
```

```python
two_histograms(x, y)
```

```python
res = stats.ttest_rel(x, y)
print('p-value (by the Pearson`s  test):', res[1])
```

```python
contingency_table = pd.DataFrame([[28, 72], [20, 80]], index = ['city', 'country'], columns = ['col', 'against'])
contingency_table
```

```python
res = stats.chi2_contingency(contingency_table)
print('p-value (by the chi-square test):', res[1])
```
